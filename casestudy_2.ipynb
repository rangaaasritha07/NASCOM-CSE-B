Task 2: Perform Data Analysis using Pandas

```python
import pandas as pd
data = {
    "Student": ["S1", "S2", "S3", "S4", "S5", "S6", "S7", "S8", "S9", "S10"],
    "Study_Hours": [2, 3, 4, 5, 6, 7, 8, 4, 6, 9],
    "Attendance": [65, 70, 75, 80, 85, 90, 92, 78, 88, 95],
    "Assignment_Score": [55, 60, 65, 72, 78, 85, 88, 70, 80, 92],
    "Previous_Score": [58, 62, 68, 70, 76, 82, 86, 72, 79, 90],
    "Final_Score": [60, 64, 70, 75, 82, 88, 91, 74, 84, 94]
}
df = pd.DataFrame(data)
print("Complete Dataset:")
print(df)
print("\nAverage Scores:")
print(df[["Study_Hours", "Attendance", "Assignment_Score",
          "Previous_Score", "Final_Score"]].mean())
highest = df.loc[df["Final_Score"].idxmax()]
print("\nStudent with Highest Final Score:")
print(highest)
lowest = df.loc[df["Final_Score"].idxmin()]
print("\nStudent with Lowest Final Score:")
print(lowest)
print("\nStudents with Attendance greater than 80%:")
print(df[df["Attendance"] > 80])
print("\nStudents with Final Score greater than 80:")
print(df[df["Final_Score"] > 80])
print("\nStatistical Summary:")
print(df.describe())
```

Task 3: Analyze Final_Score using Numpy

```python
import numpy as np
final_scores = np.array(df["Final_Score"])
print("Final Scores:")
print(final_scores)
mean = np.mean(final_scores)
print("\nMean:", mean)
median = np.median(final_scores)
print("Median:", median)
maximum = np.max(final_scores)
print("Maximum:", maximum)
minimum = np.min(final_scores)
print("Minimum:", minimum)
std = np.std(final_scores)
print("Standard Deviation:", std)
```

Task 4: Create Visualizations

```python
import pandas as pd
import matplotlib.pyplot as plt
data = {
    "Student": ["S1", "S2", "S3", "S4", "S5", "S6", "S7", "S8", "S9", "S10"],
    "Study_Hours": [2, 3, 4, 5, 6, 7, 8, 4, 6, 9],
    "Attendance": [65, 70, 75, 80, 85, 90, 92, 78, 88, 95],
    "Assignment_Score": [55, 60, 65, 72, 78, 85, 88, 70, 80, 92],
    "Previous_Score": [58, 62, 68, 70, 76, 82, 86, 72, 79, 90],
    "Final_Score": [60, 64, 70, 75, 82, 88, 91, 74, 84, 94]
}
df = pd.DataFrame(data)
plt.scatter(df["Study_Hours"], df["Final_Score"])
plt.xlabel("Study Hours")
plt.ylabel("Final Score")
plt.title("Study Hours vs Final Score")
plt.grid()
plt.show()
plt.scatter(df["Attendance"], df["Final_Score"])
plt.xlabel("Attendance (%)")
plt.ylabel("Final Score")
plt.title("Attendance vs Final Score")
plt.grid()
plt.show()
plt.hist(df["Final_Score"], bins=5, edgecolor="black")
plt.xlabel("Final Score")
plt.ylabel("Number of Students")
plt.title("Distribution of Final Scores")
plt.grid()
plt.show()
plt.bar(df["Student"], df["Final_Score"])
plt.xlabel("Student")
plt.ylabel("Final Score")
plt.title("Student-wise Final Scores")
plt.show()
```

Numerical Approximation Using Python

```python
import numpy as np
def L(x):
    return 5*x - 0.2*x**2
x = np.linspace(0, 5, 1000)
y = L(x)
area = np.trapezoid(y, x)
print("Numerical approximation:", area)
```

Correlation Analysis

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
correlation = df.corr(numeric_only=True)
print("Correlation Matrix:")
print(correlation)
print("\nStudy Hours vs Final Score:",
      correlation.loc["Study_Hours", "Final_Score"])
print("Attendance vs Final Score:",
      correlation.loc["Attendance", "Final_Score"])
print("Assignment Score vs Final Score:",
      correlation.loc["Assignment_Score", "Final_Score"])
print("Previous Score vs Final Score:",
      correlation.loc["Previous_Score", "Final_Score"])
plt.figure(figsize=(10, 7))
sns.heatmap(correlation, annot=True, cmap="coolwarm", fmt=".2f")
plt.title("Heatmap Of Correlation")
plt.show()
```

Mini Prediction Model Using Linear Algebra

```python
import numpy as np
X = df[["Study_Hours", "Attendance", "Assignment_Score"]].values
Y = df["Final_Score"].values
X = np.column_stack((np.ones(len(X)), X))
beta = np.linalg.inv(X.T @ X) @ X.T @ Y
print("Coefficients:")
print(beta)
```
