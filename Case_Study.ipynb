Complete Task 3 Code

```python
import numpy as np
A = np.array([
    [2, 1],
    [1, 3]
])
X = np.array([
    [40],
    [25]
])
Y = A @ X
print("Output vector Y:")
print(Y)
A_transpose = A.T
print("Transpose of A:")
print(A_transpose)
det_A = np.linalg.det(A)
print("Determinant of A:")
print(det_A)
if det_A != 0:
    A_inverse = np.linalg.inv(A)
    print("Inverse of A:")
    print(A_inverse)
else:
    print("Inverse does not exist")
B = np.array([
    100,
    120
])
solution = np.linalg.solve(A, B)
print("Solution of equations:")
print("x =", solution[0])
print("y =", solution[1])
```
